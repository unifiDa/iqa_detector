```python
import json
import pandas as pd
from pathlib import Path
import collections
```

```python
GEN_LOG  = Path("results.jsonl")                        # generation log: record_id, prompt, image_path, mask_path, result[url, seed]
KEPT     = Path("kept.csv")         # the images of the final dataset (column `file` = generated image name)
```

```python
log = pd.read_json(GEN_LOG, lines=True, dtype=False)
```

```python
kept = pd.read_csv(KEPT, dtype=str, usecols=["file"])
kept_files = [i_file.replace("_inpainted-0.png", "") for i_file in kept["file"]]
```

```python
kept_files[:10]
```

```python
kept_paths = [log[log.record_id == i_file].image_path.iloc[0] for i_file in kept_files]
# counter = 0
# for i_file in kept_files:
#     if i_file not in log.record_id.values:
#         counter+=1
#         print(f"{counter} - Missing record_id: {i_file}")
```

```python
# write paths to the original frames that have been collected
with open("kept_paths.json", "w") as f:
    json.dump(kept_paths, f, indent=4)
```

```python
# i film da cui sono stati estratti i frame
unique_movies = set(Path(i_path).parent.name for i_path in kept_paths)
unique_movies
```

```python
# il numero di identità coinvolte nell'inpainting se il film è lo stesso
frames = [i_name.split("_")[1] for i_name in kept_files ]
identities = collections.Counter(frames)

conteggi = collections.Counter(n for _, n in identities.items())
for k in sorted(conteggi, reverse=True):
    print(f"{k} identities: {conteggi[k] / len(identities):.2%} dei frame")
```
